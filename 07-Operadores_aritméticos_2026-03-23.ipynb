**Tarefa:**

```python
import numpy as np

# array do exercício 4 (exemplo)
array_reshape = np.array([
    [ 2.6,         2.8,        34.7286,      0.0693],
    [ 0.1386,      1.61245155,  1.67332005,  5.89309766],
    [ 0.26324893,  0.37229021,  2.6,         2.8],
    [34.7286,      0.0693,      0.1386,      1.61245155],
    [ 1.67332005,  5.89309766,  0.26324893,  0.37229021]
])
#Numero de elementos:
num_elementos = array_reshape.shape[0] * array_reshape.shape[1]
print("Número de elementos:", num_elementos)
#Média
medias = array_reshape.mean(axis=0)
novo_array = np.vstack([array_reshape, medias])
print(novo_array)
```

## 07 – Remodele (reshape) o array do exercício anterior, para que contenha 4 colunas. Converta este objeto da classe array para DataFrame. Arredonde todos os valores para até 3 casas decimais. Recupere a DataFrame.

```python
#Importar pacotes
import numpy as np
import pandas as pd

#Arrai 06
array = np.array([
    [2.6, 2.8, 34.7286, 0.0693],
    [0.1386, 1.61245155, 1.67332005, 5.89309766],
    [0.26324893, 0.37229021, 2.6, 2.8],
    [34.7286, 0.0693, 0.1386, 1.61245155],
    [1.67332005, 5.89309766, 0.26324893, 0.37229021],
    [7.8807538,   2.14942788,  7.8807538,   2.14942788]  
])
#Reshape
array_reshape = array.reshape(-1, 4)
#DataFrame
df = pd.DataFrame(array_reshape)
#Arredondar
df = df.round(3)
print(df)
```

## 08 – A última instancia desta DataFrame deve conter as médias dos valores contidos por coluna, o que foi determinado nos exercícios anteriores. Instancie um objeto contendo as médias dos valores em cada linha. Insira este objeto como uma nova coluna na DataFrame. Recupere a DataFrame.

```python
#Importação pacotes
import pandas as pd
import numpy as np
#DataFrame
df = pd.DataFrame([
    [2.6, 2.8, 34.7286, 0.0693],
    [0.1386, 1.612, 1.673, 5.893],
    [0.263, 0.372, 2.6, 2.8],
    [34.7286, 0.0693, 0.1386, 1.612],
    [1.673, 5.893, 0.263, 0.372],
    [7.48, 2.95, 7.88, 2.55]  # médias das colunas (exemplo)
])
#Médias das linhas
media_linhas = df.mean(axis=1)
df['media_linhas'] = media_linhas
print(df)
```

## 09 – Usando o método dos objetos de classe DataFrame, describe(), compare os resultados contidos na última linha e última coluna da DataFrame.

Obs.: O método dos objetos de classe DataFrame, transpose(), inverte linhas por colunas.

```python
#Importação de pacote
import pandas as pd

desc = df.describe()
ultima_linha = desc.iloc[-1]
ultima_coluna = df.iloc[:, -1].describe()

#Comparação
comparacao = ultima_linha.equals(ultima_coluna)

print("DESCRIBE")
print(desc)
print("Última linha do describe")
print(ultima_linha)
print("Última coluna (describe)")
print(ultima_coluna)
print("São iguais?")
print(comparacao)
```

## 10 – Edite o nome das colunas e dos index, para indicar onde estão as médias. Qual a média das médias?

```python
#Importação dos pacotes
import pandas as pd
#Renomear colunas
df.columns = [
    'coluna_1',
    'coluna_2',
    'coluna_3',
    'coluna_4',
    'media_linhas'
]
#Renomear index
df.index = [
    'linha_1',
    'linha_2',
    'linha_3',
    'linha_4',
    'linha_5',
    'medias_colunas'
]
#Médias
media_das_medias = df['media_linhas'].mean()
print("DATAFRAME FINAL")
print(df)

print("MÉDIA DAS MÉDIAS")
print(media_das_medias)
```
