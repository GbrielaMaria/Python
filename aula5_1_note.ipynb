```python
array3 = np.array([10, 20, 30, 40, 50])
array3
```

```python
contador1 = 0
while contador1 < len(array3):
    print(array3[contador1])
    contador1 += 1
```

```python
for i in range(len(array3)):
    print(array3[i])
```

```python
for i in enumerate(array3):
    print(i)
    print(i[1])
```
