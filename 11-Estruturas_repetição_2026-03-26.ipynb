## Instancie um objeto da classe lista contendo 5.000 valores inteiros iniciando em 0. 
Quantos são pares e quantos são impares?

## 01 – Descreva o racional da sua respos/ta:

R.: Para criar uma lista com 5.000 valores, primeiro foi criada uma lista contendo 5.000 númeor inteiros iniciando no 0 e indo até o 4999. A partir disso, os números foram classificados de acordo com a sua paridade (ímpares ou pares). Sendo que a definição para números pares foi quando divisível por 2 e o resto igual a 0; caso contrário a isso o número foi definido como ímpar. 

## 02 – Desenhe o fluxograma da resposta:

<img src="fluxogramapython4.jpg">

## 03 – Escreva o programa em Python:

```python
#Lista
lista = list(range(5000))
pares = 0
impares = 0

for n in lista:
    if n % 2 == 0:
        pares += 1
    else:
        impares += 1
print(pares)
print(impares)
```

## Importe o .csv exportado na última aula e instancie um objeto da classe DataFrame.

Dos números de letras nos genes, quais são pares e quantos são impares?

## 05 – Descreva o racional da sua resposta:

R.: Após importar o arquivo .csv gerado anteriormente, o DataFrame reconstruído continha os nomes dos genes como índice. A análise solicitada não é sobre os valores numéricos das colunas, mas sim sobre o número de caracteres de cada nome de gene. Para isso, foi calculado o comprimento de cada string presente no índice do DataFrame. Em seguida, cada tamanho é classificado como par ou ímpar utilizando a regra do exercicio anterior. E como os genes foram avaliados individualmente, bastou percorrer o índice e contar quantos apresentam comprimento par e quantos ímpar.

## 06 – Desenhe o fluxograma da resposta:

<img src="fluxogramapython5.jpg">

## 07 – Escreva o programa em Python:

```python
#Importação
import pandas as pd
#DataFrame recuperado
df = pd.read_csv("genes.csv", index_col=0, dtype=str)

genes = df.index.astype(str)
pares = sum(len(g) // 2 == 0 for g in genes)
impares = sum(len(g) // 2 != 0 for g in genes)

print(pares, impares)
```

## 08 – Substitua a estrutura de repetição e responda novamente a pergunta anterior.

```python
#Importação
import pandas as pd
#DataFrame recuperado
df = pd.read_csv("genes.csv", index_col=0, dtype=str)

genes = df.index.astype(str)
pares = sum(genes.str.len() % 2 == 0)
impares = sum(genes.str.len() % 2 != 0)

print(pares, impares)
```

## 09 – Usando a coluna de presença de hu no nome do gene, itere os index deste objeto usando a função enumerate(). Estes index podem ser usados para imprimir cada um dos valores presentes na coluna dos dois primeiros caracteres dos nomes dos genes?

```python
#Importação
import pandas as pd
#CSV
df = pd.read_csv("genes.csv", index_col=0)
genes = df.index.astype(str)
#Hu
df["has_hu"] = genes.str.contains("hu")
print(df)
for i, gene in enumerate(df.index):
    print(f"Gene {i}: {gene} -> caracteres: {gene[:2]}")
```
