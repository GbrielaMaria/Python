## 03 – Usando o método DataFrame.to_csv(‘nome_do_arquivo.csv’) exporte o objeto obtido no exercício 10 da aula anterior para um arquivo CSV. Usando o método pd.read_csv(‘nome_do_arquivo.csv’) instancie uma DataFrame. Use o valor ‘Unnamed: 0’ no parâmetro index_col, para ajustar o index. Arredonde os valores para até três casas decimais, usando o método round().
Recupere a DataFrame.

```python
#Recuperação exercicio 10:
#Importando os dados
import pandas as pd  #carregando pacote
import os   
os.listdir()   #acessando local
DataFrame = pd.read_csv('data_6.csv')  #importando dados
DataFrame.head()   #visualização 5 primeiros
#Dicionário DataFrame - dados
#Número de entradas
num_entradas = DataFrame.shape[0]   #.shape (tamanho)
print(num_entradas)
#Lista com classe das colunas
classes_colunas = list(DataFrame.dtypes)    #.dtypes (tipos)
print(classes_colunas)
#Elementos de cada classe
contagem_classes = DataFrame.dtypes.value_counts().to_dict()   #.value_counts (valores contados)
print(contagem_classes)
#Lista
colunas = [DataFrame.iloc[:,0], DataFrame.iloc[:,3], DataFrame.iloc[:, 8]]   #.iloc (localização - 1)
print(colunas)
médias = []
desvios = []
terceiro_quartil = []
#Calculando valores
for col in colunas:
    médias.append(col.mean())  #.append (adicionar) + col.mean (valor da média)
    desvios.append(col.std())  #.append (adicionar) + col.std (valor do desvio padrão)
    terceiro_quartil.append(col.quantile(0.75))  #.append (adicionar) + col.quantile (valor do quatil definido por qual parte ele vai ser 0.75, 0.25...)
#Dicionário 
resultado_dicionario = {
    'num_entradas': num_entradas,  #difinição das chaves e seu valor atribuido
    'classes_colunas': classes_colunas, 
    'contagem_classes': contagem_classes, 
    'médias': médias, 
    'desvios': desvios,
    'terceiro_quartil': terceiro_quartil
}
print(resultado_dicionario)
```

```python
#Converter o dicionário (pegando só as listas numéricas)
df_resultado = pd.DataFrame({
    'médias': resultado_dicionario['médias'],
    'desvios': resultado_dicionario['desvios'],
    'terceiro_quartil': resultado_dicionario['terceiro_quartil']
})
print(df_resultado)
#Exportar para csv
df_resultado.to_csv('resultado_df_dici.csv')
#Ler csv - indice:
df1 = pd.read_csv('resultado_df_dici.csv', index_col='Unnamed: 0')
print(df1)
#Arredondar valores
df1 = df1.round(3)
print(df1)
```

## 04 – Instancie um numpy array que apresente o mesmo formato (linhas e colunas) da DataFrame. Este array deve conter apenas inteiros uns.
Recupere o array.

Obs.: Procure na internet o método numpy.ones()

```python
#importação do pacote
import numpy as np
linhas, colunas = DataFrame.shape #lihas e colunas
array_df = np.ones((linhas, colunas), dtype=int) #.ones (preenche com 1) + dtype=int (garante inteiros)
print(array_df)
```

## 05 – Quantos dos valores contidos na DataFrame são iguais aos valores no array?

```python
#Comparação DF - ARRAY
comparação_df_array = (DataFrame.values == array_df).sum() #comparação posição por posição + .sum (contagem TRUE)
print(comparação_df_array)
```

## 06 – Quantos dos valores contidos na DataFrame são superiores aos valores no array?

```python
#Comparação DF - ARRAY
comparaçãoMAIORES_df_array = (DataFrame.select_dtypes(include='number') > 1).sum().sum() #comparação posição por posição + .sum (contagem TRUE)
print(comparaçãoMAIORES_df_array)
```

## 07 – Quantos dos valores contidos na DataFrame são iguais ou inferiores aos valores no array?

```python
#Comparação DF - ARRAY
comparaçãoMENORIGUAL_df_array = (DataFrame.select_dtypes(include='number') <= 1).sum().sum() #comparação posição por posição + .sum (contagem TRUE)
print(comparaçãoMENORIGUAL_df_array)
```

## 08 – Quantos dos valores contidos na DataFrame estão entre 0 e 1?

```python
#Comparação DF - ARRAY
comparaçãoINTERVALO_df_array = DataFrame.select_dtypes(include='number')
entre_0_1 = ((comparaçãoINTERVALO_df_array > 0) & (comparaçãoINTERVALO_df_array < 1)).sum().sum()
print(entre_0_1)
```

## 09 – Quantos dos valores das médias das linhas são inferiores ou iguais à 5?

Obs 1.: Usando o método de classe, você pode converter qualquer objeto em outra classe. Ex. obj = list(array)

Obs 2.: Assim com o objeto da classe str, listas também possuem a função count()

```python
linhas_num = DataFrame.select_dtypes(include='number') #Seleção dos dados numéricos
medias_linhas_num = linhas_num.mean(axis=1) #medias
lista1 = list(medias_linhas_num <= 5) #maior ou igual a 5
resultado = lista1.count(True) #contagem 
print(resultado)
```

## 10 – Quantos dos valores das médias das colunas são superiores à 5?

```python
colunas_num = DataFrame.select_dtypes(include='number') #Seleção dos dados numéricos
medias_colunas_num = colunas_num.mean(axis=1) #medias
lista2 = list(medias_colunas_num > 5) #maior ou igual a 5
resultado1 = lista2.count(True) #contagem 
print(resultado1)
```
