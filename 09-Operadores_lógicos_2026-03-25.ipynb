## 01 – Dados as operações abaixo, qual o resultado da operação lógica: 
p and q and r and s and t?
<br>
<br>Como os operadores lógicos podem ser usados para inverter o resultado?

<br>p = 2.0 + 3.0 / 5.0 == 10
<br>q = (p + 10.2) - 5.0 * 2.0 <= 2.9
<br>r = 10 <= p / q + 33.8 < 100
<br>s = 5 != p / (q + r)
<br>t = 33 == s * 2

```python
p = 2.0 + 3.0/5.0 == 10
print(p)
q = (p + 10.2) - 5.0 * 2.0 <= 2.9
print(q)
r = 10 <= p / q + 33.8 < 100
print(r)
s = 5 != p / (q + r)
print(s)
t = 33 == s * 2
print(t)

p and q and r and s and t 
```

```python
#Invertendo o resultado:
not (p and q and r and s and t)
```

## 02 – Usando apenas o operador or, qual seria o resultado final das operações abaixo? Como deve ser usado o not para inverter o resultado?

<br>p = ('a',1,True,'b',0,False) != ('a',True,1,'b',False,0)
<br>q = ['a',1,True,'b',0,False] != ('a',True,1,'b',False,0)
<br>r = ['a',1,True,'b',0,False] != ['a',True,1,'b',False,0]
<br>s = ['a',1,True,'b',0,False] > ['a',True,1]
<br>t = ['a',1,True,'b',0,False] == 1
<br>u = ['a',True,1] <= ['a',True,1] < ['a',1,True,'b',0,False]

```python
#Resultado original:
p = ('a',1,True,'b',0,False) != ('a',True,1,'b',False,0)
print(p)
q = ['a',1,True,'b',0,False] != ('a',True,1,'b',False,0)
print(q)
r = ['a',1,True,'b',0,False] != ['a',True,1,'b',False,0]
print(r)
s = ['a',1,True,'b',0,False] > ['a',True,1]
print(s)
t = ['a',1,True,'b',0,False] == 1
print(t)
u = ['a',True,1] <= ['a',True,1] < ['a',1,True,'b',0,False]
print(u)
```

```python
resultado_esperado = p or q or r or s or t or u
resultado_esperado
```

```python
#Not
not resultado_esperado
```

## 03 – No exercício 3 da aula 8 foi importado um arquivo .csv para um objeto da classe DataFrame pandas. Importe aqui o mesmo objeto e arredonde e edite o index como realizado no exercício 3 da aula 8. Usando operadores responda, a segunda linha é igual a terceira ou a terceira linha é igual a quarta?

```python
#Recuperação do exercicio 3:
#Carregando pacote
import pandas as pd
import os
#Ver arquivos no diretório
os.listdir()
#Importar o CSV original
DataFrame = pd.read_csv('data_6.csv')
#Criar listas para o dicionário (colunas 1, 4 e 9)
colunas = [DataFrame.iloc[:,0], DataFrame.iloc[:,3], DataFrame.iloc[:,8]]
medias = []
desvios = []
terceiro_quartil = []
for col in colunas:
    medias.append(col.mean())
    desvios.append(col.std())
    terceiro_quartil.append(col.quantile(0.75))

#Dicionário
resultado_dicionario = {
    'medias': medias,
    'desvios': desvios,
    'terceiro_quartil': terceiro_quartil
}
#DataFrame
df_resultado = pd.DataFrame(resultado_dicionario)
#Exportar para CSV
df_resultado.to_csv('resultado_df_dici.csv')
#Ler CSV - índice
df1 = pd.read_csv('resultado_df_dici.csv', index_col='Unnamed: 0')
#Arredondar
df1 = df1.round(3)
print(df1)
```

```python
# segunda linha vs terceira
linha2_eq_linha3 = df1.iloc[1].equals(df1.iloc[2])
resultado = linha2_eq_linha3 or False
print("Resultado da comparação:", resultado)
```

## 04 – Usando operadores verifique se todos os valores contidos na primeira linha são maiores ou iguais a 10.

```python
linha1 = df1.iloc[0]
resultado = (linha1 >= 10).all()
print(resultado)
```

## 05 – Usando operadores verifique se todos os valores contidos na terceira coluna estão entre 0 inclusive e 40.

```python
coluna3 = df1.iloc[:, 2]
resultado = ((coluna3 >= 0) & (coluna3 <= 40)).all()
print(resultado)
```

## 09 – Converta as duas últimas colunas do objeto DataFrame para dois objetos da classe numpy array. Algum dos valores contidos no primeiro array estão entre 2 inclusive e 5 inclusive?

```python
import numpy as np
ultimas_colunas = df1.iloc[:, -2:] #iloc[:, -2:] (pega as 2 últimas colunas)
array1 = ultimas_colunas.iloc[:, 0].to_numpy() #to_numpy() (converte para array numpy)
print(array1)
array2 = ultimas_colunas.iloc[:, 1].to_numpy()
print(array2)
resultado = ((array1 >= 2) & (array1 <= 5)).any() #(array1 >= 2) & (array1 <= 5) (intervalos) + .any() (verifica se existe pelo menos um valor)
print(resultado)
```
