## Importe o .csv utilizado no exercício 7 da aula 11 e instancie um objeto da classe DataFrame. Considere que não há uma ordem entre os genes.
Como inserir  na DataFrame uma tabela de frequência do número de caracteres no nome dos genes? Uma vez inserida, como obter um gráfico de barras deste dado?

## 01 – Descreva o racional da sua resposta:

R.: Foi calculado o número de caracteres de cada gene a partir do índice do DataFrame. Em seguida, foi construída uma tabela de frequência utilizando a contagem das ocorrências de cada tamanho. A partir disso, os dados foram representados por meio de um gráfico de barras, permitindo visualizar a distribuição do tamanho dos nomes dos genes. 

## 02 – Desenhe o fluxograma da resposta:

## 03 – Escreva o programa em Python:

```python
#Importação pacote
import pandas as pd
import matplotlib.pyplot as plt
#CSV
df = pd.read_csv("genes.csv", index_col=0, dtype=str)
#Genes
tamanho_genes = df.index.astype(str).str.len()
#Freq
freq_tamanho = tamanho_genes.value_counts().sort_index()
#DataFrame
df_freq_tamanho = freq_tamanho.to_frame(name="frequencia")
print(df_freq_tamanho)
#Gráfico
plt.figure()
df_freq_tamanho.plot(kind="bar", legend=False)
plt.title("Frequência do número de caracteres nos nomes dos genes")
plt.xlabel("Número de caracteres")
plt.ylabel("Frequência")
plt.show()
```

## Importe o .csv utilizado no exercício 7 da aula 11 e instancie um objeto da classe DataFrame. Considere que há uma ordem entre os genes.
Como inserir  na DataFrame uma tabela de frequência do número de caracteres no nome dos genes? Uma vez inserida, como obter um gráfico de barras deste dado?

## 04 – Descreva o racional da sua resposta:

R.: Considerando que há uma ordem entre os genes, foi inicialmente calculado o número de caracteres de cada gene e essa informação foi inserida como uma nova coluna no DataFrame, preservando a ordem original. Em seguida, foi construída uma tabela de frequência a partir dessa coluna, e os dados foram representados por meio de um gráfico de barras.

## 05 – Desenhe o fluxograma da resposta:

## 06 – Escreva o programa em Python:

```python
import pandas as pd
import matplotlib.pyplot as plt
#CSV
df = pd.read_csv("genes.csv", index_col=0, dtype=str)
#Coluna tamanho nome
df["tamanho_nome"] = df.index.astype(str).str.len()

print(df.head())

# ==============================
# 3. Tabela de frequência
# ==============================

freq_tamanho = df["tamanho_nome"].value_counts().sort_index()

# Converter para DataFrame
df_freq_tamanho = freq_tamanho.to_frame(name="frequencia")

print(df_freq_tamanho)

# ==============================
# 4. Gráfico de barras
# ==============================

plt.figure()

df_freq_tamanho.plot(kind="bar", legend=False)

plt.title("Frequência do número de caracteres nos genes (ordem preservada)")
plt.xlabel("Número de caracteres")
plt.ylabel("Frequência")

plt.show()
```

## 07 – Reorganize a DataFrame, removendo as colunas contendo os dois primeiros caracteres, o número de caracter 'a', e a presença de 'hu', dos genes. Renomeie as colunas de acordo com as informações. Adicione uma coluna com a informação de 'Amostra 1'.

```python
#Importação pacote
import pandas as pd
#CSV
df = pd.read_csv("genes.csv", index_col=0)
df.index = df.index.astype(str)
colunas_remover = ["primeiros_2", "num_a", "tem_hu"]
df = df.drop(columns=[c for c in colunas_remover if c in df.columns], errors='ignore')
df["primeiros_2"] = df.index.str[:2]
df["num_a"] = df.index.str.lower().str.count("a")
df["tem_hu"] = df.index.str.lower().str.contains("hu")
#Colunas
df = df.rename(columns={
    "primeiros_2": "2_primeiros_caracteres",
    "num_a": "quantidade_a",
    "tem_hu": "presenca_hu"
})

df["Amostra 1"] = "Sim"
print(df.head())
```

## 10 – Obtenha dos seus dados o gráfico abaixo, adicionando 3 replicas.

```python
#Importação pacote
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
#CSV
df = pd.read_csv("genes.csv", index_col=0)
df.index = df.index.astype(str)
#Amostras
n = len(df)
for i in range(2, 7):
    df[f"Amostra {i}"] = np.random.randint(1, 10, n)
if "Amostra 1" not in df.columns:
    df["Amostra 1"] = np.random.randint(1, 10, n)
col_amostras = [col for col in df.columns if "Amostra" in col]

df_amostras = df[col_amostras]
#Gráfico Boxplot
plt.figure()
plt.boxplot(df_amostras.values)
plt.xticks(range(1, len(col_amostras)+1), col_amostras, rotation=45)
plt.ylabel("Valores")
plt.title("Distribuição das Amostras")
plt.show()
```
