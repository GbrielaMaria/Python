# Exercício 02:

Qual as respostas das condicionais devo ter para receber uma resposta “Compre” no terminal?

<img src="./img/01-fluxograma_lampada.png" width="400" />

```python
#R: R: Para que o algoritmo retorne "Compre uma lâmpada", a condição lâmpada não está ligada (FALSE) deve ser atendida. Em seguida, a verificação do bocal deve indicar que ele não está queimado (FALSE para bocal queimado). Assim, o algoritmo conclui que o problema está na lâmpada e retorna a mensagem: "Compre uma lâmpada".
lampada_ligada = False
bocal_queimado = False

if lampada_ligada:
    print("Acenda a Luz!")
else:
    if bocal_queimado:
        print("Substitua o bocal!")
    else:
        print("Compre uma lâmpada!")
```

# Exercício 03:

Dado que a lâmpada está ligada e que o bocal está queimado, qual a linha do Python que irá imprimir no terminal a resposta ao usuário?

```python
lampada = "ligada"
bocal = "queimado"

if lampada == "ligada":
    if bocal == "queimado":
        print("Subistitua o bocal!")
    else:
        print("Compre uma nova lâmpada!")
else:
    print("Ligue a lâmpada!")
```

R.: O comando que será executado é print("Substitua o bocal!"), localizado na linha 6, pois a condição lampada == "ligada" é verdadeira e a condição bocal == "queimado" também é verdadeira
