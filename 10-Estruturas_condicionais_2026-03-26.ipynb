Caso o nome tenha entre 2 e 4 caracteres inclusive, posso saber as informações do gene? ## 07 – Descreva o racional da sua resposta:

R.: Como o exercicio pede que os genes sejam trataedos como unidades a primeira análise foi necessário reorganizar a estrutura do DataFrame. Isso foi feito por meio da transposição (.T), que converte as colunas em linhas, permitindo que cada gene passe a ser um índice (index). Essa etapa foi essencial para que a análise posterior seja feita gene a gene, e não por coluna. Além disso, como os dados são heterogeneos (números, string e valores booleanos), todos os valores foram convertidos para string para garantir consistencia e evitar conflitos de tipo durante a comparação. Assim, após a reorganização, o DataFrame foi exportado para CSV e reimportado, mantendo a estrutura corrigida. Em seguida, definiu o primeiro gene da lista como referência para aplicar a lógica condicional baseada no número de caracteres do seu nome. A partir disso, se o nome do gene tivesse entre 2 e 4 caracteres, considerava-se que há informação suficiente para acessar diretamente seus dados; se tivesse menos de 2 caracteres, a análise se restringe aos genes com nomes maiores que 4 caracteres, buscando relações apenas nesses casos; e se tivesse mais de 4 caracteres, realizou-se uma comparação global dos valores, retornando apenas os maiores. Dessa forma, o racional integra reorganização estrutural do DataFrame (transposição), padronização de tipos (string), persistência dos dados via CSV e aplicação de lógica condicional baseada em características do identificador genético.

## 09 – Escreva o programa em Python:

```python
#Importação pacote:
import pandas as pd
#Dados originais como string
genes = ["apoe", "myd88", "nfkb1", "isg15", "p53"]
dici_genes = {
    "apoe":  ["4", "ap", "1", "False"],
    "myd88": ["5", "my", "0", "False"],
    "nfkb1": ["5", "nf", "0", "False"],
    "isg15": ["5", "is", "0", "False"],
    "p53":   ["3", "p5", "0", "False"]
}
#DataFrame
df = pd.DataFrame(dici_genes)
print("DataFrame original:")
print(df)
df=df.T
df= df.astype(str)
df.index = genes
print("DataFrame corrigido")
#CSV
df.to_csv("genes.csv", index=False)
df_importado = pd.read_csv("genes.csv", dtype=str)
print("DataFrame importado:")
print(df_importado)
#Index
df_importado.index = genes
primeiro_gene = genes[0]
n_caracteres = len(primeiro_gene)
print("Primeiro gene:", primeiro_gene)
print("Número de caracteres:", n_caracteres)
if 2 <= n_caracteres <= 4:
    print("Caso 2-4 caracteres: acesso direto ao gene")
    print(df_importado.loc[primeiro_gene])
elif n_caracteres < 2:
    print("Caso <2 caracteres: genes com nome >4 caracteres")
    genes_maiores = [g for g in genes if len(g) > 4]
    print(df_importado.loc[genes_maiores])
else:
    print("Caso >4 caracteres: comparação de maior valor por coluna")
    print(df_importado.max())
```
