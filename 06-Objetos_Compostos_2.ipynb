## 10 – Importe o data.csv (dados de seguro na Flórida em 2012) para uma DataFrame. Armazene em um dicionário o número de entradas, uma lista com as classes das colunas e outra lista com o número de elementos de cada classe, três listas, contendo a média, o desvio padrão e o terceiro quartil da primeira, quarta e nona colunas . Recupere o dicionário.

```python
#Importando os dados
import pandas as pd  #carregando pacote
import os   
os.listdir()   #acessando local
DataFrame = pd.read_csv('data_6.csv')  #importando dados
DataFrame.head()   #visualização 5 primeiros
```

```python
#Dicionário DataFrame - dados
#Número de entradas
num_entradas = DataFrame.shape[0]   #.shape (tamanho)
print(num_entradas)
#Lista com classe das colunas
classes_colunas = list(DataFrame.dtypes)    #.dtypes (tipos)
print(classes_colunas)
#Elementos de cada classe
contagem_classes = DataFrame.dtypes.value_counts().to_dict()   #.value_counts (valores contados)
print(contagem_classes)
#Lista
colunas = [DataFrame.iloc[:,0], DataFrame.iloc[:,3], DataFrame.iloc[:, 8]]   #.iloc (localização - 1)
print(colunas)
médias = []
desvios = []
terceiro_quartil = []
#Calculando valores
for col in colunas:
    médias.append(col.mean())  #.append (adicionar) + col.mean (valor da média)
    desvios.append(col.std())  #.append (adicionar) + col.std (valor do desvio padrão)
    terceiro_quartil.append(col.quantile(0.75))  #.append (adicionar) + col.quantile (valor do quatil definido por qual parte ele vai ser 0.75, 0.25...)
```

```python
#Dicionário 
resultado_dicionario = {
    'num_entradas': num_entradas,  #difinição das chaves e seu valor atribuido
    'classes_colunas': classes_colunas, 
    'contagem_classes': contagem_classes, 
    'médias': médias, 
    'desvios': desvios,
    'terceiro_quartil': terceiro_quartil
}
print(resultado_dicionario)
```
