```python
lâmpada = "ligada"
bocal = "queimado" 

if lâmpada == "ligada":
    if bocal == "queimado":
        print("Substitua o bocal!")
    else:
        print("Compre uma nova lâmpada")
else:
    print("Ligue a lâmpada!")
```
