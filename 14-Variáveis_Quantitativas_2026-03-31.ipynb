## 01 – Usando o método np.random.normal(), gere valores de expressão gênica de um conjunto de 1.000 genes. Esse conjunto de ter valores up e downregulated.

```python
#Importação de pacote
import numpy as np
import pandas as pd
#Expressão
expressao = np.random.normal(loc=0, scale=1, size=1000)
#DataFrame
df = pd.DataFrame({
    "Gene": [f"Gene_{i+1}" for i in range(1000)],
    "Expressao": expressao
})
#Up ou down regulated
df["Regulacao"] = np.where(df["Expressao"] > 0, "upregulated", "downregulated")
print(df.head())
```

## 02 – Usando o método np.random.randint(), gere valores para o número de genes diferencialmente expressos em um conjunto de 1.000 pacientes. Esse conjunto deve ter valores up e downregulated.

Instancie um objeto da classe DataFrame, com os dados gerados nos dois exercícios.

```python
#Importação pacote
import numpy as np
import pandas as pd
#Exercício 1
expressao = np.random.normal(loc=0, scale=1, size=1000)
#Exercício 2
pacientes = np.random.randint(low=0, high=100, size=1000)
#DataFrame
df = pd.DataFrame({
    "Gene": [f"Gene_{i+1}" for i in range(1000)],
    "Expressao_genica": expressao,
    "Pacientes_DE": pacientes
})
#Up e down regulated
df["Regulacao"] = np.where(df["Expressao_genica"] > 0,
                           "upregulated",
                           "downregulated")
print(df.head())
```

## 03 – Categorize em grupos de 6 os dados contidos na DataFrame anterior, segundo critérios escolhidos por você. Explique os critérios.

R.: strong_down: genes muito reprimidos e com expressão abaixo do basal, significando um silenciamento gênico forte com repressão por fatores transcionais (strong_down < -2>)
moderate_down: genes com repressão clara mas não extrema, ou seja, com uma redução funcional da expressão devido a inibição parcial deles (-2 <= moderate_down < -1)
mild_down: genes com uma veriação de expressão genica pequena, geralmente associado a variação fisiologica (-1 <= mild_down < -0.5)
neutral: genes sem alteração em sua expressão, geralemnte genes estáveis associados a manutenção celular e homeostase (-0.5 <= neutral <= 0.5)
mild_up: genes com leve aumento da expressão, geralmente associado a uma ativação inicial da expressão genica relacionado a algum estimulo (0.5 < mild_up <= 1)
strong_up: genes altamente ativos, geralmente associados com uma resposta intensa como genes chaves de repsotas biologica e alta atividade transcricional.

```python
#Importação pacotes
import numpy as np
import pandas as pd
bins = [-np.inf, -2, -1, -0.5, 0.5, 1, np.inf]
labels = [
    "strong_down",
    "moderate_down",
    "mild_down",
    "neutral",
    "mild_up",
    "strong_up"
]
df["Categoria_Expressao"] = pd.cut(
    df["Expressao_genica"],
    bins=bins,
    labels=labels
)
print(df[["Gene", "Expressao_genica", "Categoria_Expressao"]].head())
```

## 04 – Defina uma função em Python que receba uma pd.Series contendo uma variável quantitativa e retorne uma DataFrame contendo a tabela de frequência.

Obtenha a tabela de frequência das variáveis criadas nos exercícios 1 e 2.

```python
#Importação pacote
import pandas as pd
#Freq
def tabela_frequencia(serie: pd.Series):
    #Frequência absoluta
    freq_abs = serie.value_counts()
    #Frequência relativa (%)
    freq_rel = serie.value_counts(normalize=True) * 100
    #DataFrame
    df_freq = pd.DataFrame({
        "Frequencia_absoluta": freq_abs,
        "Frequencia_relativa(%)": freq_rel
    })

    return df_freq
#Exercício 1:
def tabela_frequencia(serie: pd.Series):
    freq_abs = serie.value_counts()
    freq_rel = serie.value_counts(normalize=True) * 100

    df_freq = pd.DataFrame({
        "Frequencia_absoluta": freq_abs,
        "Frequencia_relativa(%)": freq_rel
    })

    return df_freq
freq_regulacao = tabela_frequencia(df["Regulacao"])
print(freq_regulacao)
#Exercício 2:
def tabela_frequencia(serie: pd.Series):
    freq_abs = serie.value_counts()
    freq_rel = serie.value_counts(normalize=True) * 100

    df_freq = pd.DataFrame({
        "Frequencia_absoluta": freq_abs,
        "Frequencia_relativa(%)": freq_rel
    })

    return df_freq
freq_regulacao = tabela_frequencia(df["Regulacao"])
print(freq_regulacao)
freq_pacientes = tabela_frequencia(df["Pacientes_DE"])
print(freq_pacientes)
```

## 08 – Gere o mesmo gráfico do exercício 06, porém, categorizado.

```python
#Importação pacote
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
#Dados exercicio 1 e 2
expressao = np.random.normal(loc=0, scale=1, size=1000)
pacientes = np.random.randint(low=0, high=100, size=1000)
df = pd.DataFrame({
    "Gene": [f"Gene_{i+1}" for i in range(1000)],
    "Expressao": expressao,
    "Pacientes_DE": pacientes
})
#Categorização
df["Regulacao"] = np.where(df["Expressao"] > 0,
                           "upregulated",
                           "downregulated")
#Histograma
plt.figure()
# Upregulated
plt.hist(df[df["Regulacao"] == "upregulated"]["Expressao"],
         bins=30,
         alpha=0.7,
         label="Upregulated")
# Downregulated
plt.hist(df[df["Regulacao"] == "downregulated"]["Expressao"],
         bins=30,
         alpha=0.7,
         label="Downregulated")
plt.title("Histograma Categorizado da Expressão Gênica")
plt.xlabel("Expressão gênica")
plt.ylabel("Frequência")
plt.legend()
plt.show()
```
